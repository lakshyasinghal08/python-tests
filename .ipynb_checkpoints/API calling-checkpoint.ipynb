```python
table = soup.find('table')
```

```python
table
```

```python
rows =[]

for row in table.find_all("tr")[1:]:
    col = table.find_all('td')

    rows.append([
        col[0].text.strip() ,
        col[1].text.strip() ,
        col[2].text.strip()
    ])
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(rows , columns=['Rank' , 'Country' ,'Population'])
```
