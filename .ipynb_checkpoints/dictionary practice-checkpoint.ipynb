```python
marks = {
    "hindi" : 89 ,
    "science" : 99 ,
    "math" : 96 ,
    "computer" : 90 ,
    "english" : 94
}
```

```python
marks
```

```python
marks.items()
```

```python
marks.keys()
```

```python
max_mark = 0
min_mark = float("inf")
total = 0
average_mark = 0
subject = ''
for key,val in marks.items():
    total += val
    if val > max_mark:
        subject = key
        max_mark = val
    if val < min_mark:
        min_mark = val
    
average_mark = total // len(marks)

print("total marks = "  , total)
print("average marks = " , average_mark)
print("max marks = " , max_mark)
print("subject of highest marks = ", subject)
print("min marks = " , min_mark)
```

```python
value = input("enter the value :-> ")
freq = {}

for i in value:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

print(freq)
```

# Take a sentence as input and count how many times each word appears.

```python
sent = input("enter the sentence :-> ")
freq = {}
lis = sent.split()

for i in lis:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1
print(freq)
```

```python
d1 = {
    'a' : 20 ,
    'b' : 20 ,
    'c' : 25
}
d2 = {
    'b' : 5 ,
    'c' : 10 ,
    'd' : 45
}
```

```python
result = {}
for key , value in d1.items():
    if key in d2.keys():
        result[key] = value
    else:
        result[key] = value
        
for key , value in d2.items():
    if key in d1.keys():
        result[key] += value
    else:
        result[key] = value
        
print(result)
```

```python
student = {}
value = int(input("enter the value :-> "))

for i in range(1,value+1):
    name = input(f"enter the name for {i} ")
    marks = int(input(f"enter the marks for {i} "))
    student[name] = marks
print(student)
```

```python
students = {}

def add_student(name , marks):
    if name in students.keys():
        print("this key is already exists")
    else:
        students[name] = marks

def search_student(name):
    if name not in students.keys():
        print("name is not exists ")
    else:
        print("marks = " , students[name])

def update_marks(name):
    if name not in students.keys():
        print("name is not present ")
    else:
        marks = int(input("enter the marks :-> "))
        students[name] = marks
        print("marks = " ,students[name])
def delete_student(name):
    if name not in students.keys():
        print("name is not present ")
    else:
        del students[name]
        print("student deleted successfully ")

def display_students():
    print(students)
```

```python
add_student('aman' ,  95)
add_student('lakshya' , 96)
add_student('virat' , 100)
add_student('kohli' , 100)
display_students()
```
