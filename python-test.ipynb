# Q1. Find Maximum — 2 Marks
### Find the largest number from a list.
### Input: numbers = [10, 20, 5, 30]
### Output: 30
### Condition: Don't use max().

```python
numbers = [10, 20, 5, 30]
maxi = 0

for i in numbers:
    if i > maxi:
        maxi = i
print("maximum marks = " , maxi)
```

# Q3. Create Dictionary — 2 Marks
## Create a dictionary using two lists.
 ## names = ["A", "B", "C"]
 ## marks = [70, 80, 90]
 ## Output: {"A": 70, "B": 80, "C": 90}


```python
names = ["A", "B", "C"]
marks = [70, 80, 90]
```

```python
result = {}

for i,j in zip(names , marks):
    result[i] = j

print(result)
```

# Q4. Sum of Digits

```python
num = int(input("enter the number :-> "))
sum = 0

while num > 0:
    digit = num % 10
    sum += digit
    num //= 10

print("sum = "  , sum)
```

# Q6. Check Palindrome —

```python
num = int(input("enter the number :-> "))
original = num
rev = 0

while num > 0:
    digit = num % 10
    rev = rev*10 + digit
    num //= 10

if original == rev:
    print("palindrome")
else:
    print("not palindrome")
```

# Q12. Print Pattern — 2 Marks
Write a program to print:
### *
### **
### ***
### ****
Input: 4

```python
i = 0
while i < 4:
    j = 0
    while j < 4:
        if i >= j:
            print("*",end=" ")
        else:
            print(" ",end= " ")
        j += 1
    print()
    i += 1
```

# Q18. Average and Maximum 

```python
data = {"a": 5, "b": 10, "c": 15}
avg = 0
maxi = 0

for i, j in data.items():
    avg += j
    if j > maxi:
        maxi = j

avg = avg / len(data)

print("Average =", avg)
print("Maximum =", maxi)
```

# Q20. Employee & Developer

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
class Developer(Employee):
    def __init__(self, name, salary, language):
        Employee.__init__(self, name, salary)  
        self.language = language

    def display(self):
        print("Name = ", self.name)
        print("Salary = ", self.salary)
        print("Language = ", self.language)

name = "lakshya"
salary = 900000
language = "ML"
a = Developer(name, salary, language)
a.display()
```
